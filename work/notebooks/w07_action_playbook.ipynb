```python
# This cell is for CODE (numbers, a query, a check).

from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# 1. LOCATE THE FLYRANK FILES
# ============================================================

SEARCH_ROOT = Path("/content")

# Find the raw dataset
raw_matches = list(
    SEARCH_ROOT.rglob("content_refresh_anonymized.csv")
)

if not raw_matches:
    raise FileNotFoundError(
        "Could not find content_refresh_anonymized.csv anywhere under /content.\n"
        "Please upload or clone the FlyRank repository first."
    )

RAW_PATH = raw_matches[0]

# Repository root:
# .../data/raw/content_refresh_anonymized.csv
#                    ↑
# RAW_PATH.parent.parent.parent = repository root
ROOT = RAW_PATH.parent.parent.parent

PROCESSED_PATH = ROOT / "processed"
OUTPUTS_PATH = ROOT / "outputs"

COMMITTED_QUEUE_PATH = OUTPUTS_PATH / "refresh_queue_sample.csv"
FULL_QUEUE_PATH = OUTPUTS_PATH / "refresh_queue.csv"
MODEL_PREDICTIONS_PATH = PROCESSED_PATH / "model_predictions.csv"


print("=" * 60)
print("FLYRANK PATH CHECK")
print("=" * 60)

print(f"Repository root : {ROOT}")
print(f"Raw dataset     : {RAW_PATH}")
print(f"Raw exists      : {RAW_PATH.exists()}")
print(f"Processed dir   : {PROCESSED_PATH}")
print(f"Outputs dir     : {OUTPUTS_PATH}")

print("=" * 60)


# ============================================================
# 2. FIND THE VALIDATED QUEUE
# ============================================================

if FULL_QUEUE_PATH.exists():
    source_queue_path = FULL_QUEUE_PATH

elif MODEL_PREDICTIONS_PATH.exists():
    source_queue_path = MODEL_PREDICTIONS_PATH

elif COMMITTED_QUEUE_PATH.exists():
    source_queue_path = COMMITTED_QUEUE_PATH

else:
    # Search the entire environment in case the files are
    # located somewhere unexpected.
    queue_matches = []

    for filename in [
        "refresh_queue.csv",
        "model_predictions.csv",
        "refresh_queue_sample.csv"
    ]:
        queue_matches.extend(
            SEARCH_ROOT.rglob(filename)
        )

    if queue_matches:
        source_queue_path = queue_matches[0]
    else:
        raise FileNotFoundError(
            "\nCould not find a validated refresh queue.\n\n"
            "Expected one of:\n"
            "  - outputs/refresh_queue.csv\n"
            "  - processed/model_predictions.csv\n"
            "  - outputs/refresh_queue_sample.csv\n\n"
            "The raw dataset alone is not enough for this notebook.\n"
            "You need the Week-6/validated queue output."
        )


print(f"Using queue: {source_queue_path}")
print("=" * 60)


# ============================================================
# 3. LOAD THE QUEUE
# ============================================================

queue = pd.read_csv(source_queue_path)

print(f"Queue shape: {queue.shape}")
print()


# ============================================================
# 4. CHECK REQUIRED COLUMNS
# ============================================================

required_columns = {
    "content_id",
    "client_id",
    "best_model_probability",
    "suggested_action",
    "final_reason_codes",
    "impressions_90d",
    "sessions_90d",
    "avg_position",
    "ctr",
    "content_age_days",
    "days_since_last_update",
    "word_count"
}

missing_columns = required_columns.difference(queue.columns)

if missing_columns:
    raise ValueError(
        "Validated queue is missing required columns: "
        f"{sorted(missing_columns)}"
    )


# ============================================================
# 5. CREATE PREDICTION SCORE
# ============================================================

score_column = (
    "final_refresh_score"
    if "final_refresh_score" in queue.columns
    else "best_model_probability"
)

queue["prediction_score"] = pd.to_numeric(
    queue[score_column],
    errors="coerce"
)

queue["model_probability"] = pd.to_numeric(
    queue["best_model_probability"],
    errors="coerce"
)

queue = queue.dropna(
    subset=[
        "prediction_score",
        "model_probability"
    ]
).copy()


# ============================================================
# 6. RANK THE CONTENT
# ============================================================

queue = queue.sort_values(
    [
        "prediction_score",
        "model_probability"
    ],
    ascending=False
).reset_index(drop=True)

queue["priority_rank"] = np.arange(
    1,
    len(queue) + 1
)


# ============================================================
# 7. CREATE PRIORITY BANDS
# ============================================================

queue["priority_band"] = np.select(
    [
        queue["priority_rank"] <= 50,
        queue["priority_rank"] <= 500
    ],
    [
        "top_50",
        "next_450"
    ],
    default="remaining_review"
)


# ============================================================
# 8. SUPPORTED REASON CODES
# ============================================================

def supported_reason_codes(row):

    reasons = []

    # Model decline risk
    if row["model_probability"] >= 0.65:
        reasons.append(
            "model_decline_risk"
        )

    # Visible opportunity
    if (
        row["impressions_90d"] >= 500
        and row["model_probability"] >= 0.50
    ):
        reasons.append(
            "visible_model_opportunity"
        )

    # CTR review candidate
    if (
        row["impressions_90d"] >= 500
        and 0 < row["avg_position"] <= 20
        and row["ctr"] < 0.50
    ):
        reasons.append(
            "ctr_review_candidate"
        )

    # Engagement review candidate
    engagement_rate = row.get(
        "engagement_rate",
        0
    )

    scroll_rate = row.get(
        "scroll_rate",
        0
    )

    if (
        row["sessions_90d"] >= 30
        and (
            0 < engagement_rate < 30
            or 0 < scroll_rate < 30
        )
    ):
        reasons.append(
            "engagement_review_candidate"
        )

    if not reasons:
        return "general_review"

    return "|".join(reasons)


queue["supported_reason_codes"] = queue.apply(
    supported_reason_codes,
    axis=1
)


# ============================================================
# 9. NORMALIZE RECOMMENDED ACTION
# ============================================================

queue["recommended_action"] = (
    queue["suggested_action"]
    .replace({
        "refresh": "refresh_for_review"
    })
)


# ============================================================
# 10. SELECT USEFUL COLUMNS
# ============================================================

useful_columns = [
    "priority_rank",
    "priority_band",
    "content_id",
    "client_id",
    "prediction_score",
    "model_probability",
    "recommended_action",
    "supported_reason_codes",
    "impressions_90d",
    "sessions_90d",
    "avg_position",
    "ctr",
    "content_age_days",
    "days_since_last_update",
    "word_count"
]


# Optional columns
for optional_column in [
    "content_type",
    "main_intent",
    "competition_level",
    "trend_direction"
]:

    if optional_column in queue.columns:
        useful_columns.append(
            optional_column
        )


# ============================================================
# 11. CREATE FINAL ACTION QUEUE
# ============================================================

action_queue = queue[
    useful_columns
].copy()


# ============================================================
# 12. FINAL OUTPUT
# ============================================================

try:
    relative_source = source_queue_path.relative_to(ROOT)
except ValueError:
    relative_source = source_queue_path


print()
print("=" * 60)
print("ACTION PLAYBOOK QUEUE")
print("=" * 60)

print(
    f"Source       : {relative_source}"
)

print(
    f"Rows ranked  : {len(action_queue):,}"
)

print(
    f"Top 50       : "
    f"{(action_queue['priority_band'] == 'top_50').sum():,}"
)

print(
    f"Next 450     : "
    f"{(action_queue['priority_band'] == 'next_450').sum():,}"
)

print(
    f"Remaining    : "
    f"{(action_queue['priority_band'] == 'remaining_review').sum():,}"
)

print("=" * 60)

display(
    action_queue.head(10)
)
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*


This is a prioritization and review aid for SEO and content teams. Start with the highest-ranked items, inspect the supplied signals, and decide whether a page deserves editorial investigation.

The model ranks or flags items in this observed dataset; it does not decide what to publish and it does not establish that refreshing a page will improve traffic, clicks, rankings, or engagement. The data is a trailing 90-day snapshot, so scores can be affected by seasonality, client mix, measurement gaps, and changes after the snapshot. A high score is evidence for review priority, not proof that an action is correct. Recommendations are most defensible within the data population and validation setup used for the model.

```python
# This cell is for CODE (numbers, a query, a check).
review_rules = pd.DataFrame(
    {
        "rule": [
            "Start with high-priority rows, then inspect the underlying signals.",
            "Reject or investigate rows with sparse, stale, or implausible measurements.",
            "Require a human owner and rollback plan for any proposed change.",
            "Never automate publishing, deletion, redirects, or irreversible SEO changes.",
        ],
        "status": ["required", "required", "required", "no-go"],
    }
)
display(review_rules)

no_go_actions = {
    "automatic_publish": False,
    "automatic_delete": False,
    "automatic_redirect": False,
    "automatic_irreversible_seo_change": False,
}
assert not any(no_go_actions.values())
print("Human review is required before every action.")
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

## 3. Human review + the no-go list

A reviewer should confirm that the item is still live and in scope, check the page and search intent, verify that the observed traffic and position signals are plausible, and record the editorial decision. Reviewers should reject or investigate a recommendation when the page has changed recently, the data is sparse or stale, the recommendation conflicts with current business or legal guidance, or the proposed change would require evidence not present in this dataset.

**No-go list:** do not automatically publish, delete, redirect, canonicalize, change internal-link structure, alter metadata at scale, or make any other irreversible content or SEO change from a model prediction alone. The model must not choose claims, medical/legal/financial advice, brand-sensitive wording, or compliance decisions. A human owns the final action, rationale, and rollback plan.

## 4. Monitoring / retrain triggers

These checks are lightweight research safeguards, not a production monitoring service. Before reusing the queue, compare the current data with the validation snapshot and inspect whether the recommendation mix still looks plausible.

Revalidate the model when ranking quality degrades on newly labeled items, when score distributions or key feature distributions shift, when missingness or coverage changes, when new labeled data becomes available, or when search and traffic behavior changes materially. Retraining is appropriate only after the label definition, feature availability, and leakage checks are reviewed again. A changed search environment is a reason to investigate and revalidate, not evidence that the model caused the change.

```python
# This cell is for CODE (numbers, a query, a check).
monitoring_snapshot = pd.DataFrame(
    {
        "check": [
            "queue row count",
            "missing prediction score",
            "duplicate content_id",
            "probability range",
            "top-50 share with model probability >= 0.65",
        ],
        "value": [
            len(action_queue),
            int(action_queue["prediction_score"].isna().sum()),
            int(action_queue["content_id"].duplicated().sum()),
            f"{action_queue['model_probability'].min():.3f} to {action_queue['model_probability'].max():.3f}",
            float((action_queue.head(50)["model_probability"] >= 0.65).mean()),
        ],
    }
)
assert monitoring_snapshot.loc[1, "value"] == 0
assert monitoring_snapshot.loc[2, "value"] == 0
assert action_queue["model_probability"].between(0, 1).all()
display(monitoring_snapshot)

revalidation_triggers = [
    "precision@K or other agreed ranking metric degrades on newly labeled items",
    "score, feature, missingness, or client-mix distributions shift materially",
    "new labeled data changes the target definition or available population",
    "search, traffic, measurement, or platform behavior changes materially",
]
print("Revalidation triggers:")
for trigger in revalidation_triggers:
    print(f"- {trigger}")
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

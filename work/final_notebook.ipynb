# Week 8 Showcase — 5-Minute Demo

## 1. Question — 45 seconds

**Can we identify content that is showing signs of decline using historical content and search-performance signals?**

The goal of this project was to investigate declining content using anonymized FlyRank production search and content-performance data and determine whether machine learning could help prioritize pages that may need attention.

---

## 2. Method — 1 minute

The analysis followed these steps:

* Explored the anonymized FlyRank content-performance dataset.
* Investigated signals including:

  * impressions
  * sessions
  * CTR
  * average search position
  * content age
  * historical trend direction
* Defined content decline as the target prediction problem.
* Built features from historical content and search-performance signals.
* Trained classification models to identify potentially declining content.
* Evaluated the model using a validation approach designed to reduce overly optimistic results.
* Reviewed the results and limitations before making recommendations.

## 4. One Honest Result — 1 minute

The final Random Forest model achieved **65.7% accuracy vs 63.3%** majority-class baseline (+2.4 percentage points)

This result suggests that the historical content and search-performance features contain some useful information for identifying potentially declining content.

However, the improvement over the baseline is modest. Therefore, the model should not be presented as a fully automated production decision-maker. Instead, it should be treated as a prioritization signal that can support human content decisions and help identify pages that deserve further review.

---
## 5. One Recommendation — 1 minute 15 seconds

Use the analysis as a **content-prioritization tool** rather than an automatic content-rewriting system.

A practical workflow would be:

1. Identify pages with strong decline or weak click-capture signals.
2. Prioritize pages that have meaningful search impressions but relatively low CTR.
3. Use the **CTR Opportunity Score** to create a ranked review queue.
4. Consider content age, historical performance, search position, and other available signals before making a decision.
5. Send high-priority pages to a content or SEO specialist for human review.
6. After updating selected pages, monitor their search performance to determine whether the intervention improved results.

The key recommendation is to use ML and opportunity scoring to **help FlyRank decide which pages deserve attention first**, while keeping the final content decision with a human reviewer.
